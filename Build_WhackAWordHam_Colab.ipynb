# 🐹 Whack-A-Word-Ham: Free Google Colab Android APK Builder
Follow the steps below in order to compile your APK in ~8-10 minutes and download it directly to your PC.

> **Note:** OpenSSL source tarballs are automatically pre-cached from GitHub Releases in Step 3 to prevent `openssl.org` HTTP 502/504 gateway timeouts.

### Step 1: Clone your GitHub Repository

```python
!rm -rf ScramTheWord
!git clone https://github.com/irai1a/ScramTheWord.git
%cd ScramTheWord
```

### Step 2: Install Android Build Dependencies

```python
!sudo apt update
!sudo apt install -y openjdk-17-jdk autoconf libtool pkg-config zlib1g-dev libncurses5-dev libncursesw5-dev libtinfo5 cmake libffi-dev libssl-dev libsqlite3-dev
!pip install --upgrade pip
!pip install "cython<3.0" buildozer virtualenv
```

### Step 3: Pre-Cache OpenSSL & Build the Android APK

```python
# Pre-cache OpenSSL archives from GitHub Releases to prevent openssl.org 502/504 timeout
!mkdir -p /root/.buildozer/android/packages/openssl
!curl -fsSL -o /root/.buildozer/android/packages/openssl/openssl-3.3.1.tar.gz https://github.com/openssl/openssl/releases/download/openssl-3.3.1/openssl-3.3.1.tar.gz
!touch /root/.buildozer/android/packages/openssl/.mark-openssl-3.3.1.tar.gz
!curl -fsSL -o /root/.buildozer/android/packages/openssl/openssl-3.0.12.tar.gz https://github.com/openssl/openssl/releases/download/openssl-3.0.12/openssl-3.0.12.tar.gz
!touch /root/.buildozer/android/packages/openssl/.mark-openssl-3.0.12.tar.gz
!curl -fsSL -o /root/.buildozer/android/packages/openssl/openssl-1.1.1w.tar.gz https://github.com/openssl/openssl/releases/download/OpenSSL_1_1_1w/openssl-1.1.1w.tar.gz
!touch /root/.buildozer/android/packages/openssl/.mark-openssl-1.1.1w.tar.gz

# Run Buildozer Android Debug Build
!buildozer -v android debug
```

### Step 4: Download the Compiled APK to your Computer

```python
import glob
from google.colab import files

apk_files = glob.glob('bin/*.apk')
if apk_files:
    print(f'Downloading {apk_files[0]}...')
    files.download(apk_files[0])
else:
    print('No APK found in bin/ directory. Please check Step 3 logs above.')
```

### Optional Step 5: Build Google Play Release AAB (App Bundle)

```python
!buildozer -v android release

import glob
from google.colab import files
aab_files = glob.glob('bin/*.aab')
if aab_files:
    print(f'Downloading {aab_files[0]}...')
    files.download(aab_files[0])
```
